```python
### Homework

# TODO: make a corpus object for both files
# TODO: distinguish between spanish and english
# TODO: how to compare the english vs spanish
# TODO: model time in a text √
# TODO: make a chapter class (Brandon can show you how to link them)
# TODO: emdashes in words
# TODO: add in the other symbols besides ángel
# TODO: output the format for network analysis
# TODO: do the network analysis

class Text(object):
    def __init__(self, fn):
        self.filename = fn
        self.raw_text = self.read_in_text()
        self.chapters = [Chapter(chapter_text) for chapter_text in self.raw_text.split('Capítulo')]
        self.sentences = self.flatten([chapter.clean_tokenized_sentences for chapter in self.chapters])
        self.tokens = self.flatten([chapter.sans_stopwords_tokens for chapter in self.chapters])
        self.paragraphs = [paragraph for paragraph in self.raw_text.split('\n\n')]
        self.pages = [page for page in re.split(r'\n[0-9]+\n', self.raw_text)]
        self.ángel_sentences = [' '.join(sentence) for sentence in self.sentences if 'ángel' in sentence]
        print('i am a text')
    
    def read_in_text(self):
        with open(self.filename, 'r') as file_in:
            raw_text = file_in.read()
        return raw_text
    
    def flatten(self, list_of_lists):
        # removes one level of lists
        return [item for sublist in list_of_lists for item in sublist]
    
class Chapter(object):
    def __init__(self, raw_text):
        self.raw_text = raw_text
        # building up our stopwords list
        self.stopwords = stopwords.words('Spanish')
        self.stopwords.extend(string.punctuation)
        # add some stuff in that was left out of the stopwords list
        extra_stopwords = ['“','”']
        self.stopwords.extend(extra_stopwords)
        # editing down the stopwords list to keep certain words in. these are the words we DO care about.
        remove_stopwords = ['muy', 'durante']
        for word in remove_stopwords:
            self.stopwords.remove(word)
        self.raw_sentences = nltk.sent_tokenize(self.raw_text)
        self.raw_tokenized_sentences = [nltk.word_tokenize(sent) for sent in self.raw_sentences]
        self.raw_tokens = nltk.word_tokenize(self.raw_text)
        self.sans_stopwords_tokens = [token.lower() for token in self.raw_tokens if token not in self.stopwords]
        self.clean_tokenized_sentences = []
        for sent in self.raw_tokenized_sentences:
            clean_sentence = []
            for word in sent:
                if word not in self.stopwords:
                    clean_sentence.append(word)
            self.clean_tokenized_sentences.append(clean_sentence)
            
        self.fq = nltk.FreqDist(self.sans_stopwords_tokens)

    

# function template
#     def function_name(self):
#         some stuff
#         return something

# To test, make sure you rerun this block and the blocks where you instantiate the object
```
