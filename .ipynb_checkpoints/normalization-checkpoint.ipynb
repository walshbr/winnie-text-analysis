```python
lower_tokens = [token.lower() for token in raw_tokens] # makes all text lowercase

# remove stopwords
no_stopwords_tokens = []

nltk.download('stopwords')
stopw = stopwords.words('Spanish')

for token in lower_tokens:
    if token is not stopw:
        no_stopwords_tokens.append(token) 
print(no_stopwords_tokens)
```

```python
### TODO: Textcleaning

# TODO: Normalize whitespace (done by tokenizing √)
# TODO: Remove dashes between syllables (needs to be done by hand)
# TODO: Spelling correction (best done in the file with regex)

### Homework

# TODO: make a corpus object for both files
# TODO: distinguish between spanish and english
# TODO: how to compare the english vs spanish
# TODO: model time in a text
# TODO: make a chapter class (Brandon can show you how to link them)

class Text(object):
    def __init__(self, fn):
        self.filename = fn
        self.raw_text = self.read_in_text()
        self.chapters = [Chapter(chapter_text) for chapter_text in self.raw_text.split('Capítulo')]        
        print('i am a text')
        self.tokens = [chapter.clean_tokens for chapter in self.chapters]

    
    def read_in_text(self):
        with open(self.filename, 'r') as file_in:
            raw_text = file_in.read()
        return raw_text
    
class Chapter(object):
    def __init__(self, raw_text):
        self.raw_text = raw_text
        self.stopwords = stopwords.words('Spanish')
        #TODO: add punctuation to the list of stopwords using string.punctuation 
        # (hint: string.punctuation is a string and you need to add it to the stopwords list)
        self.raw_sentences = nltk.sent_tokenize(self.raw_text)
        self.raw_tokenized_sentences = [nltk.word_tokenize(sent) for sent in self.raw_sentences]
        self.raw_tokens = nltk.word_tokenize(self.raw_text)
        self.clean_tokens = [token.lower() for token in self.raw_tokens if token not in self.stopwords]
        self.clean_tokenized_sentences = []
        for sent in self.raw_tokenized_sentences:
            clean_sentence = []
            for word in sent:
                if word not in self.stopwords:
                    clean_sentence.append(word)
            self.clean_tokenized_sentences.append(clean_sentence)
        self.fq = nltk.FreqDist(self.clean_tokens)

    

# function template
#     def function_name(self):
#         some stuff
#         return something

# To test, make sure you rerun this block and the blocks where you instantiate the object
```
